```python
import numpy as np

def selesaikan_spltv(A, b):
    """
    Menyelesaikan Sistem Persamaan Linear Tiga Variabel (SPLTV)
    bentuk Ax = b menggunakan metode Eliminasi Gauss.
    """
    n = len(b)
    # Menggabungkan matriks koefisien A dan vektor b menjadi augmented matrix
    augmented = np.hstack((A, b.reshape(-1, 1)))

    # 1. Eliminasi Maju (Forward Elimination)
    for i in range(n):
        # Partial pivoting untuk akurasi numerik
        max_row = i + np.argmax(np.abs(augmented[i:, i]))
        if abs(augmented[max_row, i]) < 1e-10:
            raise ValueError("Matriks singular atau tidak memiliki solusi unik.")

        # Tukar baris jika diperlukan
        augmented[[i, max_row]] = augmented[[max_row, i]]

        # Buat elemen di bawah diagonal utama menjadi nol
        for j in range(i + 1, n):
            faktor = augmented[j, i] / augmented[i, i]
            augmented[j, i:] -= faktor * augmented[i, i:]

    # 2. Substitusi Balik (Back Substitution)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (augmented[i, -1] - np.dot(augmented[i, i+1:n], x[i+1:n])) / augmented[i, i]

    return x

if __name__ == "__main__":
    # Contoh Kasus SPLTV:
    # 2x + y + z = 4
    # x + 3y + 2z = 6
    # x + 1y + 5z = 7

    A = np.array([
        [2, 1, 1],
        [1, 3, 2],
        [1, 1, 5]
    ], dtype=float)

    b = np.array([4, 6, 7], dtype=float)

    print("Koefisien Matriks A:\n", A)
    print("Vektor Konstanta b:", b)

    solusi = selesaikan_spltv(A, b)
    print("\nHasil Solusi SPLTV:")
    print(f"x = {solusi[0]:.4f}")
    print(f"y = {solusi[1]:.4f}")
    print(f"z = {solusi[2]:.4f}")
```
